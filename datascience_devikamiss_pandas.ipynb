loc and iloc
loc=accessing

iloc=silicing

```python
data1=[12,13,14,15,16]    #loc used to accessing elements from customized index
s=pd.Series(data1,index=["a","b","c","d","e"])
print(s.loc["a"])
```

```python
data1=[12,13,14,15,16]     #iloc is interger based accessing that is default index
s=pd.Series(data1)
print(s.iloc[1])
```

```python
data1=[12,13,14,15,16]    #loc silicing (ivide letters ayond postion start from 1)
s=pd.Series(data1,index=["a","b","c","d","e"])
print(s.loc["a":"b"])
```

```python
data1=[12,13,14,15,16]     #iloc silicing(numbers position start from 0)
s=pd.Series(data1)
print(s.iloc[1:3])
```

```python
import pandas as pd
sales=pd.Series([1200,850,1500,950,2000],index=["monday","tuesday","wednesday","thursday","friday"])
print(sales)
```

```python
import pandas as pd
sales=pd.Series([1200,850,1500,950,2000],index=["monday","tuesday","wednesday","thursday","friday"])
print(sales.max())
print(sales.min())
print(sales.sum())
print(sales[sales<1000])
```

```python
import pandas as pd
data=[20,20,30,30,50,60,60,80]
ar=pd.Series(data)
print(ar)
print(ar.unique()) #remove repeated number it use(unique)
print(ar.nunique())#  count unique number
print(ar.value_counts())   # how many times a number repeated
```
